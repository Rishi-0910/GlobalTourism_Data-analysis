# Data Visualization_05

### Importing libraries

```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns
import warnings 
warnings.filterwarnings('ignore')
```

### Importing Datasets

```python
Clean_DataSet=pd.read_csv(r"C:\Users\Rishik Reddy\OneDrive\Desktop\Travelnfo\DataSets\ProcessedData\Merged_Tourism_Dataset.csv")
Clean_DataSet.head()
```

## 1) Revenue per Tourist Ranking

```python
top_spending = Clean_DataSet.groupby("Country Name")["AvgSpendingPerTourist"].mean().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))
sns.barplot(x=top_spending.values, y=top_spending.index)

plt.title("Top 10 Countries by Average Spending per Tourist")
plt.xlabel("Average Spending per Tourist")
plt.ylabel("Country")
plt.show()
```

## Graph summary

* **Qatar Highest Spending** – Qatar records the highest spending per tourist. <br>

* **High-Value Destinations** – Libya and Tuvalu also show very high spending levels. <br>

* **Strong Tourism Value** – Countries like Australia and Luxembourg generate high revenue per tourist. <br>

* **Revenue Strategy** – Some countries earn more by increasing spending per tourist rather than increasing tourist numbers. <br>

## 2) Arrivals vs Expenditure

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=Clean_DataSet,
    x="Arrivals",
    y="Expenditure",
    alpha=0.6
)

plt.title("Tourist Arrivals vs Tourism Expenditure")
plt.xlabel("Arrivals")
plt.ylabel("Expenditure")
plt.show()
```

## Graph summary

* **Positive Relationship** – The scatter plot shows that tourism expenditure increases as tourist arrivals increase. <br>

* **High Tourism Countries** – Some countries attract very large numbers of tourists and generate extremely high tourism spending. <br>

* **Low Tourism Markets** – Many countries appear in the lower range with fewer tourists and lower expenditure. <br>

* **Economic Impact** – The visualization indicates that tourist arrivals play a major role in increasing tourism revenue. <br>

## 3) Tourism Market Share

```python
remove_list = [
"Post-demographic dividend",
"IDA & IBRD total",
"IBRD only",
"Euro area",
"Early-demographic dividend",
"Late-demographic dividend",
"IDA total",
"Small states",
"Fragile and conflict affected situations",
"Pre-demographic dividend",
]
Clean_DataSet=Clean_DataSet[~Clean_DataSet["Country Name"].isin(remove_list)]
```

```python
market_share = Clean_DataSet.groupby("Country Name")["Expenditure"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(8,8))

plt.pie(
    market_share.values,
    labels=market_share.index,
    autopct="%1.1f%%",
    startangle=140
)

plt.title("Top 10 Countries Tourism Market Share (Expenditure)")
plt.show()
```

## Graph summary 

* **United States Dominates** – The United States holds the largest share of tourism expenditure among the top countries. <br>

* **European Tourism Power** – Germany and France contribute a large portion of global tourism spending. <br>

* **Other Strong Markets** – Countries like Japan, Australia, and Korea also contribute significant tourism expenditure. <br>

* **Market Concentration** – A few countries control a large share of global tourism spending. <br>

##  4) Top 10 Highest Tourism Arrivals

```python
remove_list = [
"Post-demographic dividend",
"IDA & IBRD total",
"IBRD only",
"Euro area",
"Early-demographic dividend",
"Late-demographic dividend",
"IDA total",
"Small states",
"Fragile and conflict affected situations",
"Pre-demographic dividend",
"IDA only",
]
Clean_DataSet=Clean_DataSet[~Clean_DataSet["Country Name"].isin(remove_list)]
```

```python
arrival_variation = Clean_DataSet.groupby("Country Name")["Arrivals"].std().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,6))

sns.barplot(
    x=arrival_variation.values,
    y=arrival_variation.index
)

plt.title("Countries with Highest Tourism Arrival Variability")
plt.xlabel("Arrival Variability (Std Dev)")
plt.ylabel("Country")
plt.show()
```

## Graph summary

* **United States Highest Variability** – The U.S. shows the highest fluctuation in tourist arrivals across years. <br>

* **Growing Tourism Markets** – China and France also experience large changes in tourist arrivals. <br>

* **Moderate Variability** – Countries like Italy, Mexico, and Poland show moderate changes in tourism demand. <br>

* **Tourism Sensitivity** – Variability indicates that tourism is affected by global events, economy, and travel trends. <br>

## 5) Tourism Growth vs Spending Efficiency

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=Clean_DataSet,
    x="TourismGrowth",
    y="AvgSpendingPerTourist",
    alpha=0.6
)

plt.title("Tourism Growth vs Average Spending per Tourist")
plt.xlabel("Tourism Growth")
plt.ylabel("Average Spending per Tourist")
plt.show()
```

## Graph summary 

* **Weak Relationship** – Tourism growth does not strongly determine how much tourists spend. <br>

* **Outliers Present** – Some countries show very high spending per tourist even with low tourism growth. <br>

* **Major Cluster** – Most countries fall within moderate growth and moderate spending levels. <br>

* **Different Tourism Models** – Some destinations focus on high-spending tourists instead of large tourist volumes. <br>
